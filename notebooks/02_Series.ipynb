**Objetivo:** entender a estrutura de uma `Series`: criação, índice personalizado, média, `idxmax`, filtros, acesso aos dados e fatiamento.

## 1. Criando uma Series

Uma `Series` é uma **coluna de dados com um índice**. Quando não informamos o índice,
o pandas numera as posições a partir de `0`.

```python
import pandas as pd

dados = [10, 20, 30, 40, 50]
series = pd.Series(dados)
series
```

### Series com índice personalizado

Podemos trocar os números do índice por rótulos — aqui, os meses do ano.

```python
vendas_mensais = pd.Series(
    [2000, 2500, 3000, 2800, 3500, 4000, 4500, 5000, 4800, 5200, 6000, 7000],
    index=["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"],
)
vendas_mensais
```

## 2. Média de vendas

`mean()` calcula a média de todos os valores da Series.

```python
media_vendas = vendas_mensais.mean()
media_vendas
```

## 3. Mês com maior venda

`max()` devolve o **maior valor**; `idxmax()` devolve o **rótulo do índice** onde ele está.

```python
mes_maior_venda = vendas_mensais.idxmax()
print(f"Maior venda: {vendas_mensais.max()} em {mes_maior_venda}")
```

## 4. Vendas acima da média (filtro)

A comparação `vendas_mensais > media_vendas` gera uma Series de `True`/`False`.
Usada entre colchetes, ela **filtra** só os meses onde o resultado é `True`.

```python
vendas_mensais > media_vendas
```

```python
vendas_acima_media = vendas_mensais[vendas_mensais > media_vendas]
vendas_acima_media
```

## 5. Acessando dados

Na **lista** Python, acessamos pela posição. Na **Series**, existem dois jeitos explícitos:

- `.iloc[posição]` → pela posição (0, 1, 2…), igual à lista.
- `.loc[rótulo]` → pelo rótulo do índice (`"Jan"`, `"Fev"`…).

```python
print("Lista  :", dados[0], dados[1], dados[2])
print("Series :", series.iloc[0], series.iloc[1], series.iloc[2])
```

```python
print("Pelo rótulo :", vendas_mensais.loc["Jan"])
print("Pela posição:", vendas_mensais.iloc[0])
```

## 6. Fatiamento (slicing)

`[1:3]` pega da posição 1 **até antes** da 3 (o 3 não entra).

> Atenção: com `.loc` e rótulos, o fim **entra** no resultado (`"Fev":"Abr"` inclui abril).

```python
print("Lista:", dados[1:3])
```

```python
series.iloc[1:3]
```

```python
vendas_mensais.loc["Fev":"Abr"]
```

## 7. Series com índice de letras

O índice pode ser qualquer rótulo, e os valores podem ser decimais (`float`).

```python
ex1 = pd.Series([0.25, 0.75, 1.0], index=["a", "b", "c"], name="Series do Exemplo 1")
ex1
```

```python
print(ex1["a"], ex1["b"], ex1["c"])
```
