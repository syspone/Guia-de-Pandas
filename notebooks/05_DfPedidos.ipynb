**Objetivo:** ler o arquivo `Dataset/Pedidos.csv` com pandas e exibir o `DataFrame`, montando o caminho do arquivo com `pathlib`.

## 1. Importações

- `pathlib.Path` monta caminhos de arquivo de forma segura em qualquer sistema operacional.
- `pandas` é importado com o apelido `pd`, a convenção usada por toda a comunidade.

```python
from pathlib import Path

import pandas as pd
```

## 2. Opções de exibição

Por padrão o pandas esconde colunas com `...` quando a tabela é larga.
As opções abaixo mostram **todas as colunas** sem truncar.

> As opções precisam ser definidas **antes** de exibir o DataFrame, senão não têm efeito.
> Não precisamos liberar todas as linhas: o pandas já mostra até 60 linhas e esta base tem 43.

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

## 3. Caminho do arquivo

Nos scripts `.py` usávamos `Path(__file__)` para achar a pasta do projeto.
Em um notebook **não existe `__file__`**, então procuramos a pasta `Dataset`
a partir da pasta atual (`Path.cwd()`), subindo um nível de cada vez.

```python
# Raiz do projeto: a primeira pasta (atual ou acima) que contém "Dataset".
# Assim o notebook funciona aberto tanto da pasta notebooks/ quanto da raiz.
RAIZ = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "Dataset").is_dir())
DATA_DIR = RAIZ / "Dataset"
ARQUIVO = DATA_DIR / "Pedidos.csv"

if not ARQUIVO.exists():
    raise FileNotFoundError(f"Arquivo não encontrado: {ARQUIVO}")

ARQUIVO.relative_to(RAIZ)
```

## 4. Leitura do CSV

`pd.read_csv` transforma o arquivo em um `DataFrame` (uma tabela com linhas e colunas).

- `sep=","` → separador das colunas no arquivo.
- `encoding="utf-8"` → garante que acentos como em *Paraíba* sejam lidos corretamente.

```python
df = pd.read_csv(ARQUIVO, sep=",", encoding="utf-8")

print(f"Arquivo: {ARQUIVO.name} ({ARQUIVO.stat().st_size} bytes)")
print(f"Linhas: {len(df)} | Colunas: {list(df.columns)}")
```
