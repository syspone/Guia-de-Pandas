## 1. Importações

```python
from pathlib import Path

import pandas as pd
```

## 2. Opções de exibição

Mostramos **todas as colunas** (são 20), mas **não** liberamos todas as linhas:
esta base tem mais de 100 mil registros e exibi-los de uma vez travaria o notebook.
Para olhar os dados usamos `head()`, que mostra só o começo.

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

## 3. Caminho do arquivo e colunas de data

As colunas de data são informadas em `parse_dates` para já entrarem como
`datetime` (data de verdade), e não como texto.

```python
# Raiz do projeto: a primeira pasta (atual ou acima) que contém "Dataset".
# Assim o notebook funciona aberto tanto da pasta notebooks/ quanto da raiz.
RAIZ = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "Dataset").is_dir())
DATA_DIR = RAIZ / "Dataset"
ARQUIVO = DATA_DIR / "GazPrice.csv"

COLUNAS_DATA = ["DATA INICIAL", "DATA FINAL"]
```

## 4. Carregando o dataset

Guardamos a leitura numa função: fica fácil reaproveitar e o erro é claro se o arquivo sumir.

```python
def carregar_dataset(caminho: Path) -> pd.DataFrame:
    """Lê o CSV de preços de combustíveis e devolve um DataFrame."""
    if not caminho.exists():
        raise FileNotFoundError(f"Arquivo não encontrado: {caminho}")

    return pd.read_csv(caminho, sep=",", encoding="utf-8", parse_dates=COLUNAS_DATA)


df_gaz = carregar_dataset(ARQUIVO)
print(f"Arquivo: {ARQUIVO.name} ({ARQUIVO.stat().st_size:,} bytes)")
```

## 5. Tipo do objeto

Confirma que o resultado da leitura é um `pandas.DataFrame`.

```python
type(df_gaz)
```

## 6. Dimensões com `shape`

`shape` é um **atributo** (sem parênteses) e devolve uma tupla `(linhas, colunas)`.

```python
linhas, colunas = df_gaz.shape
print(f"Shape: {df_gaz.shape}")
print(
    f"O DataFrame possui {linhas} linhas/observações/registros "
    f"e {colunas} colunas/atributos/variáveis."
)
```

## 7. Estrutura com `info()`

`info()` é um **método** (com parênteses): mostra o tipo de cada coluna,
quantos valores não nulos existem e quanta memória o DataFrame ocupa.
Repare que as colunas de data aparecem como `datetime64`.

```python
df_gaz.info()
```

## 8. Amostra com `head()`

`head(n)` retorna apenas as `n` primeiras linhas (o padrão é 5).

```python
df_gaz.head()
```
