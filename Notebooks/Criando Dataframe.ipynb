```python
# ==> Criando Dataframe
personagens_df = pd.DataFrame({
    'nome': ['Cristian', 'Jessica Fernanda', 'Bruna'],
    'idade' : [39, 36, 30],
    'peso': [96.5, 50.5, 60.6],
    'eh_jedi': [True, False, True]
})
```

```python
personagens_df
```

💡 Diferença rápida:
*   Método (.rename()): Executa uma ação ou transformação nos dados e exige parênteses.
* Atributo (ex: .shape, .columns): Apenas retorna uma propriedade ou informação sobre o objeto, sem executar uma ação, e não usa parênteses.

```python
personagens_df_renomeado = personagens_df.rename(
    columns={
        'nome': 'Nome Completo',                        # Renomeando coluna com método rename()
        'idade': 'Idade'
    },
    inplace=True                                        # inplace é uma argumento que modifica o dataframe
)
```
